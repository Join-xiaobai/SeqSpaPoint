```python
import pickle
import numpy as np

def read_dta_Y_file(y_file_path: str):
    """
    读取DTA数据集无后缀Y文件（pickle序列化格式），并新增NaN探查功能
    :param y_file_path: Y文件完整路径
    :return: 亲和力矩阵（numpy数组）
    """
    # 1. 验证文件是否存在
    try:
        file = open(y_file_path, "rb")
        file.close()
    except FileNotFoundError:
        raise FileNotFoundError(f"未找到Y文件，请确认路径正确：{y_file_path}")
    
    # 2. 核心读取逻辑（保留你原有的论文同款方案）
    try:
        affinity_matrix = pickle.load(open(y_file_path, "rb"), encoding="latin1")
        print("✅ Y文件读取成功！")
    except Exception as e:
        raise Exception(f"❌ Y文件读取失败，错误信息：{e}\n提示：请确认文件路径正确，且文件为DTA数据集的pickle序列化文件")

    # 3. 新增：NaN全面探查（基于numpy数组，适配亲和力矩阵）
    print("\n" + "="*50)
    print("📊 KIBA Y文件（pickle格式）NaN探查结果")
    print("="*50)
    # 3.1 全局NaN统计
    total_nan = np.isnan(affinity_matrix).sum()
    total_elements = affinity_matrix.size
    nan_percent = (total_nan / total_elements) * 100 if total_elements > 0 else 0
    print(f"1. 全局NaN情况：")
    print(f"   - NaN总数量：{total_nan}")
    print(f"   - NaN占比：{nan_percent:.6f}%")
    print(f"   - 矩阵总元素数：{total_elements}")

    # 3.2 按行（药物/配体）统计NaN
    row_nan_count = np.isnan(affinity_matrix).sum(axis=1)
    has_nan_rows = np.where(row_nan_count > 0)[0]
    print(f"\n2. 按行（药物/配体）统计：")
    print(f"   - 包含NaN的行数：{len(has_nan_rows)}")
    print(f"   - 最多NaN的行索引：{np.argmax(row_nan_count)}，对应NaN数量：{np.max(row_nan_count)}")
    if len(has_nan_rows) <= 20:
        print(f"   - 含NaN的行索引：{has_nan_rows.tolist()}")
    else:
        print(f"   - 前20个含NaN的行索引：{has_nan_rows[:20].tolist()}")

    # 3.3 按列（靶点/蛋白质）统计NaN
    col_nan_count = np.isnan(affinity_matrix).sum(axis=0)
    has_nan_cols = np.where(col_nan_count > 0)[0]
    print(f"\n3. 按列（靶点/蛋白质）统计：")
    print(f"   - 包含NaN的列数：{len(has_nan_cols)}")
    print(f"   - 最多NaN的列索引：{np.argmax(col_nan_count)}，对应NaN数量：{np.max(col_nan_count)}")
    if len(has_nan_cols) <= 20:
        print(f"   - 含NaN的列索引：{has_nan_cols.tolist()}")
    else:
        print(f"   - 前20个含NaN的列索引：{has_nan_cols[:20].tolist()}")

    # 3.4 验证矩阵基本信息（保留你原有的验证逻辑）
    print(f"\n4. 亲和力矩阵基本信息：")
    print(f"   - 数据类型：{type(affinity_matrix)}")
    print(f"   - 矩阵形状：{affinity_matrix.shape}")
    print(f"   - 元素数据类型：{affinity_matrix.dtype}")
    print(f"   - 最小值：{np.nanmin(affinity_matrix):.4f}（忽略NaN计算）")
    print(f"   - 最大值：{np.nanmax(affinity_matrix):.4f}（忽略NaN计算）")
    print(f"   - 平均值：{np.nanmean(affinity_matrix):.4f}（忽略NaN计算）")
    print(f"   - 前5行前5列数据：\n{affinity_matrix[:5, :5]}")
    print("="*50 + "\n")

    return affinity_matrix

# ------------------- 使用示例（无需修改，直接运行） -------------------
if __name__ == "__main__":
    # 替换为你的Y文件实际路径（无后缀）
    y_file_path = "./Y"  # 如：/root/me/data/dta/kiba/Y 或 r"D:\dta\kiba\Y"
    
    # 读取Y文件并自动完成NaN探查
    affinity_matrix = read_dta_Y_file(y_file_path)
```
